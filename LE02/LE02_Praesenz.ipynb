### Aufgabe 1: Funktionen und ein eigenes Modul (ca. 15 min)

In LE01 haben Sie die Dauer einer Session mehrfach berechnet und den Code dafür kopiert. Das ändern wir jetzt. *(Selbststudium: Abschnitte 6, 7 und 9)*

Ergänzen Sie in der Zelle unten die drei Funktionen:

a) `dauer_min(start, ende)` gibt die Dauer zwischen zwei Uhrzeiten im Format `HH:MM` in Minuten zurück, auch über Mitternacht (`"23:15"` bis `"00:25"` sind 70 Minuten).

b) `wochentag(datum)` gibt zu einem ISO-Datum wie `"2026-10-05"` das deutsche Kürzel `"Mo"` zurück.

c) `normalisiere_fach(fach)` entfernt Leerzeichen am Rand und vereinheitlicht die Schreibweise (`" mathe "` → `"Mathe"`).

d) Prüfen Sie Ihre Funktionen mit den Tests in der Zelle darunter. Erst wenn `Alle Tests bestanden.` erscheint, geht es weiter.

<details><summary>💡 Tipp zu a)</summary>

Die Lösung für das Mitternachtsproblem steht im Selbststudium (Abschnitt 9): Liegt `schluss` vor `beginn`, addieren Sie `timedelta(days=1)`. Die Differenz `schluss - beginn` ist ein `timedelta`. Mit `.total_seconds()` erhalten Sie die Sekunden, mit `// 60` daraus die Minuten. Denken Sie an `int(...)`.

</details>

<details><summary>💡 Tipp zu b)</summary>

`date.fromisoformat(datum)` macht aus dem String ein Datum. Dessen Methode `.weekday()` liefert `0` für Montag bis `6` für Sonntag. Diese Zahl ist genau der passende Index für die Liste `WOCHENTAGE`.

</details>

<details><summary>💡 Tipp zu c)</summary>

Zwei String-Methoden genügen: `strip()` entfernt Leerzeichen am Rand, `title()` schreibt den ersten Buchstaben groß und den Rest klein. Methoden lassen sich verketten: `text.strip().title()`.

</details>

````python
from datetime import date, datetime, timedelta

WOCHENTAGE = ["Mo", "Di", "Mi", "Do", "Fr", "Sa", "So"]


def dauer_min(start: str, ende: str) -> int:
    """Dauer zwischen zwei Uhrzeiten im Format HH:MM in Minuten, auch über Mitternacht."""
    beginn = datetime.strptime(start, "%H:%M")
    schluss = datetime.strptime(ende, "%H:%M")
    if schluss > beginn:
        schluss += timedelta(days=1)

    dauer = schluss - beginn
    return dauer


def wochentag(datum: str) -> str:
    """Deutsches Wochentagskürzel zu einem ISO-Datum, z. B. '2026-10-05' -> 'Mo'."""
    datum = datetime.strptime(datum, "%Y-%m-%d")
    return datum.strftime("%B")


def normalisiere_fach(fach: str) -> str:
    """Entfernt Leerzeichen am Rand und vereinheitlicht die Schreibweise, z. B. ' mathe ' -> 'Mathe'."""
    liste = [element.strip() for element in fach]
    return liste 
````
