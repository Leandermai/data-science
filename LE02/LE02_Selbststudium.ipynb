## 1 Listen
---

Eine **Liste** ist eine geordnete, veränderliche Folge von Werten. Sie wird mit eckigen Klammern geschrieben. Indexing und Slicing funktionieren genau wie bei Strings.

````python
dauern = [95, 40, 105, 35, 110]   # Dauer von Kims Sessions in Minuten

print(dauern[0], dauern[-1])       # erstes und letztes Element
print(dauern[1:3])                 # Slicing
print(len(dauern), sum(dauern), min(dauern), max(dauern))
print(40 in dauern)
````

Anders als Strings sind Listen **veränderlich** (*mutable*). Die wichtigsten Methoden:

| Methode | Wirkung |
|---|---|
| `liste.append(x)` | `x` am Ende anhängen |
| `liste.extend(andere)` | alle Elemente einer anderen Liste anhängen |
| `liste.insert(i, x)` | `x` an Position `i` einfügen |
| `liste.remove(x)` | erstes Vorkommen von `x` entfernen |
| `liste.pop()` / `liste.pop(i)` | letztes Element bzw. Element an Position `i` entfernen und zurückgeben |
| `liste.sort()` | Liste **an Ort und Stelle** sortieren (gibt `None` zurück) |
| `sorted(liste)` | **neue**, sortierte Liste zurückgeben, das Original bleibt unverändert |

````python
dauern.append(90)
print(dauern)

sortiert = sorted(dauern)
print(sortiert)
print(dauern)          # unverändert

dauern.sort(reverse=True)
print(dauern)          # jetzt absteigend sortiert
````

Listen können beliebige Werte enthalten, auch andere Listen. Eine **Liste von Listen** ist eine einfache Darstellung einer Tabelle:

````python
tabelle = [
    ["2026-10-05", "Mathe", 95],
    ["2026-10-05", "Statistik", 40],
    ["2026-10-06", "Mathe", 35],
]
print(tabelle[1])       # zweite Zeile
print(tabelle[1][2])    # zweite Zeile, dritte Spalte
````

### Achtung: Variablen sind Namen, keine Kopien

In Python ist eine Variable ein **Name**, der auf ein Objekt verweist. `b = a` erzeugt bei einer Liste deshalb keine Kopie. Beide Namen verweisen danach auf **dieselbe** Liste:

````python
a = [1, 2, 3]
b = a
b.append(4)
print(a)          # a hat sich mit verändert!
print(a is b)     # dasselbe Objekt
````

Eine echte Kopie erhalten Sie mit `a.copy()` oder `list(a)`. Diese Eigenheit wird uns bei pandas wieder begegnen. Merken Sie sich: Veränderliche Objekte werden **geteilt**, nicht kopiert.

````python
a = [1, 2, 3]
b = a.copy()
b.append(4)
print(a, b)
````

## 2 Tupel
---

Ein **Tupel** ist wie eine Liste, aber **unveränderlich**. Es wird mit runden Klammern geschrieben. Tupel eignen sich für feste Datensätze, deren Teile zusammengehören, z. B. eine Koordinate oder eine Session:

````python
session = ("2026-10-05", "Mathe", 95)
datum, fach, dauer = session        # Unpacking kennen Sie aus LE01
print(fach, dauer)
````

````python
session[2] = 100
````

Tupel begegnen Ihnen vor allem an drei Stellen:

* Eine Funktion gibt **mehrere Werte** zurück, z. B. `divmod(95, 60)` → `(1, 35)`.
* Als **Schlüssel in Dictionaries**, denn nur unveränderliche Werte dürfen Schlüssel sein (siehe nächster Abschnitt).
* Beim **Iterieren** über Paare, z. B. mit `zip` oder `dict.items()`.

````python
stunden, minuten = divmod(95, 60)
print(f"{stunden} h {minuten} min")
````

## 3 Dictionaries
---

Ein **Dictionary** (`dict`) ordnet **Schlüsseln** (*keys*) **Werte** (*values*) zu. Statt über eine Position greifen Sie über einen Namen zu. Damit ist ein Dictionary die natürliche Darstellung **eines Datensatzes**, also einer Zeile mit benannten Spalten:

````python
session = {"datum": "2026-10-05", "fach": "Mathe", "dauer_min": 95, "fokus": 4}

print(session["fach"])
session["ort"] = "Bibliothek"          # neues Schlüssel-Wert-Paar
session["fokus"] = 5                   # Wert überschreiben
print(session)
print("notiz" in session)              # prüft die Schlüssel
````

Der Zugriff auf einen Schlüssel, der nicht existiert, führt zu einem `KeyError`. Mit `.get()` geben Sie stattdessen einen Standardwert an:

````python
session["notiz"]
````

````python
session.get("notiz", "")
````

Mit `.keys()`, `.values()` und `.items()` iterieren Sie über ein Dictionary:

````python
for schluessel, wert in session.items():
    print(f"{schluessel:<10} {wert}")
````

### Das Zähl- und Summiermuster

Das wichtigste Einsatzmuster für Dictionaries in der Datenanalyse: **Werte nach einem Schlüssel gruppieren und aufsummieren.** Statt einer Variablen pro Fach verwenden Sie ein Dictionary, in dem das Fach der Schlüssel ist:

````python
sessions = [("Mathe", 95), ("Statistik", 40), ("Mathe", 35), ("Programmieren", 110), ("Statistik", 60)]

lernzeit = {}
for fach, dauer in sessions:
    lernzeit[fach] = lernzeit.get(fach, 0) + dauer

lernzeit
````

`lernzeit.get(fach, 0)` liefert die bisherige Summe oder `0`, wenn das Fach noch nicht vorkommt. Ein neues Fach landet so automatisch im Dictionary. Das Muster funktioniert mit beliebig vielen Kategorien.

Und für **zwei** Dimensionen? Schlüssel dürfen auch Tupel sein, z. B. `("Mo", "Mathe")`. Genau damit lösen Sie in der Präsenz die Frage aus LE01.

💡 Die Standardbibliothek bietet für dieses Muster auch fertige Werkzeuge: [`collections.Counter`](https://docs.python.org/3/library/collections.html#collections.Counter) zum Zählen und [`collections.defaultdict`](https://docs.python.org/3/library/collections.html#collections.defaultdict) für Standardwerte.

## 4 Sets und Überblick
---

Ein **Set** ist eine ungeordnete Menge **ohne Duplikate**. Sets sind ideal, um eindeutige Werte zu finden und Mengen zu vergleichen:

````python
faecher_roh = ["Mathe", "Statistik", "Mathe", "Programmieren", "Statistik"]
print(set(faecher_roh))

woche1 = {"Mathe", "Statistik", "Programmieren"}
woche2 = {"Mathe", "Datenbanken"}
print(woche1 | woche2)    # Vereinigung: in mindestens einer Woche
print(woche1 & woche2)    # Schnittmenge: in beiden Wochen
print(woche1 - woche2)    # Differenz: nur in Woche 1
````

💡 Ein leeres Set schreiben Sie als `set()`, denn `{}` ist ein leeres Dictionary.

| Struktur | Schreibweise | geordnet | veränderlich | Duplikate | typischer Einsatz |
|---|---|---|---|---|---|
| Liste | `[1, 2, 2]` | ja | ja | ja | Folge gleichartiger Werte, z. B. alle Dauern |
| Tupel | `(1, "a")` | ja | nein | ja | fester Datensatz, mehrere Rückgabewerte, Dict-Schlüssel |
| Dictionary | `{"a": 1}` | ja (Einfügereihenfolge) | ja | Schlüssel eindeutig | Datensatz mit Namen, Gruppieren und Zählen |
| Set | `{1, 2}` | nein | ja | nein | eindeutige Werte, Mengenvergleiche |

## 5 Iterieren und Comprehensions
---

### Nützliche Helfer für Schleifen

* `enumerate(folge, start=1)` liefert zu jedem Element eine laufende Nummer.
* `zip(a, b)` läuft über zwei (oder mehr) Folgen parallel.
* `sorted(folge, key=…, reverse=True)` sortiert nach einem selbst gewählten Kriterium.

````python
faecher = ["Mathe", "Statistik", "Programmieren"]
minuten = [130, 100, 110]

for nr, fach in enumerate(faecher, start=1):
    print(nr, fach)

for fach, m in zip(faecher, minuten):
    print(f"{fach:<15}{m:>5} min")
````

Um ein Dictionary nach seinen **Werten** zu sortieren, übergeben Sie `sorted` eine Funktion als `key`. Für solche kurzen Einweg-Funktionen gibt es **Lambda-Ausdrücke**: `lambda paar: paar[1]` ist eine namenlose Funktion, die von einem Paar das zweite Element zurückgibt.

````python
lernzeit = {"Mathe": 130, "Statistik": 100, "Programmieren": 110}
sorted(lernzeit.items(), key=lambda paar: paar[1], reverse=True)
````

### Comprehensions

Mit einer **Comprehension** erzeugen Sie eine neue Liste, ein Set oder ein Dictionary in einer einzigen, gut lesbaren Zeile. Das Muster lautet: *„Ausdruck für jedes Element aus einer Folge, optional mit einer Bedingung“*.

````python
dauern = [95, 40, 105, 35, 110]

in_stunden = [d / 60 for d in dauern]                 # umrechnen
lange = [d for d in dauern if d >= 60]                # filtern
print(in_stunden)
print(lange)

faecher = {f.strip().title() for f in ["Mathe", "mathe ", "Statistik"]}   # Set-Comprehension
print(faecher)

stunden = {fach: m / 60 for fach, m in lernzeit.items()}                # Dict-Comprehension
print(stunden)
````

Die Comprehension `[d / 60 for d in dauern]` entspricht dieser Schleife:

```python
in_stunden = []
for d in dauern:
    in_stunden.append(d / 60)
```

💡 Comprehensions eignen sich für einfache Umformungen und Filter. Sobald Sie mehrere Bedingungen verschachteln müssen, ist eine normale Schleife meist lesbarer.

## 6 Funktionen
---

Funktionen kapseln Logik, die Sie mehrfach brauchen, unter einem Namen. In LE01 haben Sie die Berechnung der Dauer mehrfach kopiert. Das ist fehleranfällig: Ein Fehler müsste an allen Stellen korrigiert werden. Eine Funktion definieren Sie mit `def`:

````python
def formatiere_dauer(minuten: int, kurz: bool = False) -> str:
    """Formatiert eine Dauer in Minuten als lesbaren Text.

    Beispiele:
        formatiere_dauer(95)            -> '1 h 35 min'
        formatiere_dauer(95, kurz=True) -> '1:35'
    """
    stunden, rest = divmod(minuten, 60)
    if kurz:
        return f"{stunden}:{rest:02d}"
    return f"{stunden} h {rest} min"


print(formatiere_dauer(95))
print(formatiere_dauer(95, kurz=True))
````

Die Bestandteile im Überblick:

* **Parameter** stehen in der Klammer. `kurz: bool = False` hat einen **Standardwert** und ist deshalb optional.
* Beim Aufruf können Sie Argumente per Position oder per Name übergeben: `formatiere_dauer(95, kurz=True)`. Benannte Argumente (*Keyword Arguments*) machen Aufrufe lesbarer.
* `return` gibt das Ergebnis zurück. Ohne `return` liefert eine Funktion `None`.
* Die **Type Hints** (`minuten: int`, `-> str`) dokumentieren, welche Typen erwartet werden. Python prüft sie nicht, Editoren nutzen sie aber für Hinweise.
* Der **Docstring** in dreifachen Anführungszeichen beschreibt, was die Funktion tut. Er erscheint bei `help(formatiere_dauer)` und bei `Shift + Tab`.

Mehrere Rückgabewerte geben Sie als Tupel zurück:

````python
def statistik(werte: list[float]) -> tuple[float, float, float]:
    """Gibt Minimum, Mittelwert und Maximum einer Liste zurück."""
    return min(werte), sum(werte) / len(werte), max(werte)


kleinster, mittel, groesster = statistik([95, 40, 105, 35, 110])
print(kleinster, mittel, groesster)
````

### Gültigkeitsbereich (Scope)

Variablen, die **innerhalb** einer Funktion entstehen, sind **lokal**. Sie existieren nur während des Aufrufs. Eine Funktion sollte ihre Eingaben über Parameter erhalten und ihr Ergebnis mit `return` zurückgeben, statt auf Variablen außerhalb zuzugreifen oder diese zu verändern. Solche Funktionen sind leicht zu verstehen, zu testen und wiederzuverwenden.

````python
def verdopple(x):
    ergebnis = 2 * x
    return ergebnis

verdopple(21)
print(ergebnis)       # existiert außerhalb der Funktion nicht
````

⚠️ **Klassischer Stolperstein:** Verwenden Sie niemals eine Liste oder ein Dictionary als Standardwert (`def f(werte=[])`). Der Standardwert wird nur **einmal** erzeugt und dann bei jedem Aufruf wiederverwendet. Schreiben Sie stattdessen `def f(werte=None)` und legen Sie die Liste in der Funktion an.

### Funktionen testen mit `assert`

`assert bedingung` bricht mit einem `AssertionError` ab, wenn die Bedingung nicht erfüllt ist. Damit können Sie schnell prüfen, ob eine Funktion für bekannte Fälle das Richtige liefert, gerade auch für **Sonderfälle**:

````python
assert formatiere_dauer(95) == "1 h 35 min"
assert formatiere_dauer(60) == "1 h 0 min"
assert formatiere_dauer(5, kurz=True) == "0:05"
print("Alle Tests bestanden.")
````

### Pfade mit `pathlib`

Das Modul `pathlib` stellt Pfade als Objekte dar. Mit `/` setzen Sie Pfade zusammen, und das funktioniert unter Linux, macOS und Windows gleichermaßen:

````python
from pathlib import Path

daten = Path("data")
datei = daten / "lernsessions.csv"

print(datei)
print(datei.exists(), datei.name, datei.stem, datei.suffix)
print(sorted(daten.glob("*.csv")))       # alle CSV-Dateien im Ordner
````

`Path("data")` ist ein **relativer Pfad**. Er bezieht sich auf das aktuelle Arbeitsverzeichnis, bei Notebooks ist das der Ordner des Notebooks (`Path.cwd()`). Verwenden Sie in Ihrem Projekt immer relative Pfade. Absolute Pfade wie `C:\Users\kim\...` funktionieren nur auf Ihrem Rechner.

### Textdateien lesen

Dateien öffnen Sie mit `open()` innerhalb eines `with`-Blocks. Am Ende des Blocks wird die Datei automatisch geschlossen, auch wenn ein Fehler auftritt. Geben Sie immer die Kodierung `encoding="utf-8"` an, sonst werden Umlaute je nach Betriebssystem falsch gelesen.

````python
with open(datei, encoding="utf-8") as f:
    for nr, zeile in enumerate(f):
        print(zeile.rstrip())       # rstrip entfernt den Zeilenumbruch am Ende
        if nr == 4:
            break
````

Für kleine Dateien geht es auch kürzer: `datei.read_text(encoding="utf-8")` liefert den gesamten Inhalt als einen String.

### CSV-Dateien: warum nicht einfach `split(",")`?

In LE01 haben Sie Zeilen mit `split` zerlegt. Bei echten CSV-Dateien geht das schief, sobald ein Wert selbst ein Trennzeichen enthält. Solche Werte stehen dann in Anführungszeichen:

````python
zeilen = datei.read_text(encoding="utf-8").splitlines()
for zeile in zeilen:
    if '"' in zeile:
        print(zeile)
        print(len(zeile.split(",")), "Teile statt 8:", zeile.split(","))
        break
````

Das Modul `csv` beherrscht diese Regeln. Am praktischsten ist der `DictReader`: Er verwendet die Kopfzeile als Schlüssel und liefert jede Zeile als Dictionary. Das Ergebnis ist eine **Liste von Dictionaries**, also genau die zeilenorientierte Tabelle aus Abschnitt 3.

````python
import csv

with open(datei, encoding="utf-8", newline="") as f:
    sessions = list(csv.DictReader(f))

print(len(sessions), "Sessions")
sessions[0]
````

⚠️ **Alle Werte sind Strings**, auch `fokus_1bis5` und `pausen_anzahl`, und fehlende Werte sind leere Strings `""`. Die Umwandlung in passende Typen ist Ihre Aufgabe (siehe Abschnitt 10).

💡 Deutsche Excel-Exporte verwenden oft `;` als Trennzeichen. Dann schreiben Sie `csv.DictReader(f, delimiter=";")`. Das Argument `newline=""` empfiehlt die Dokumentation für CSV-Dateien, damit Zeilenumbrüche innerhalb von Werten korrekt behandelt werden.

### CSV-Dateien schreiben

Mit `csv.DictWriter` schreiben Sie eine Liste von Dictionaries zurück in eine Datei. `mkdir(exist_ok=True)` legt den Zielordner an, falls er noch nicht existiert:

````python
uebersicht = [
    {"fach": "Mathe", "minuten": 130},
    {"fach": "Statistik", "minuten": 100},
]

ziel = Path("ergebnisse")
ziel.mkdir(exist_ok=True)

with open(ziel / "uebersicht.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["fach", "minuten"])
    writer.writeheader()
    writer.writerows(uebersicht)

print((ziel / "uebersicht.csv").read_text(encoding="utf-8"))
````

### JSON-Dateien

**JSON** (*JavaScript Object Notation*) ist das Standardformat von Web-APIs und vielen Datenexporten, z. B. von Spotify, Google oder Smartwatches. JSON-Strukturen entsprechen direkt Python-Datenstrukturen:

| JSON | Python |
|---|---|
| Objekt `{"a": 1}` | `dict` |
| Array `[1, 2]` | `list` |
| String `"text"` | `str` |
| Zahl `42`, `4.5` | `int`, `float` |
| `true`, `false` | `True`, `False` |
| `null` | `None` |

Kims Smartwatch exportiert die Schlafdaten als JSON. So lesen Sie die Datei ein und erkunden ihre Struktur:

````python
import json

with open(daten / "schlaf.json", encoding="utf-8") as f:
    schlaf = json.load(f)

print(type(schlaf))
print(schlaf.keys())
print(schlaf["device"], schlaf["units"])
print(len(schlaf["sleepSessions"]), "Nächte")
schlaf["sleepSessions"][0]
````

Eine JSON-Datei ist oft **verschachtelt**: Die Nächte liegen als Liste unter dem Schlüssel `"sleepSessions"`, jede Nacht ist ein Dictionary, und darin steckt unter `"stages"` noch ein Dictionary. Auf verschachtelte Werte greifen Sie Schritt für Schritt zu:

````python
erste_nacht = schlaf["sleepSessions"][0]
print(erste_nacht["stages"]["deep"], "Minuten Tiefschlaf")
````

Zum Schreiben verwenden Sie `json.dump(objekt, f, indent=2, ensure_ascii=False)`. `indent` sorgt für eine lesbare Einrückung, `ensure_ascii=False` erhält Umlaute.

## 9 Datum und Zeit
---

Das Modul `datetime` stellt Typen für Datum (`date`), Uhrzeit (`time`), Zeitpunkt (`datetime`) und Zeitspanne (`timedelta`) bereit. Damit rechnen Sie viel sicherer als mit selbst zerlegten Strings.

````python
from datetime import date, datetime, timedelta

tag = date.fromisoformat("2026-10-05")
print(tag, tag.isoweekday())            # 1 = Montag
print(tag.strftime("%d.%m.%Y"))         # Ausgabe im deutschen Format
print(tag + timedelta(days=7))          # eine Woche später
````

**Parsen und Formatieren:** `datetime.strptime(text, format)` wandelt einen String in einen Zeitpunkt um (*parse*), `strftime(format)` wandelt zurück in einen String (*format*). Die wichtigsten Formatcodes:

| Code | Bedeutung | Beispiel |
|---|---|---|
| `%Y` | Jahr, vierstellig | `2026` |
| `%m` | Monat, zweistellig | `10` |
| `%d` | Tag, zweistellig | `05` |
| `%H`, `%M` | Stunde (24 h), Minute | `08`, `50` |

````python
beginn = datetime.strptime("05.10.2026 08:50", "%d.%m.%Y %H:%M")
ende = datetime.fromisoformat("2026-10-05T10:25")      # ISO-Format braucht kein Formatmuster
dauer = ende - beginn

print(type(dauer), dauer)
print(dauer.total_seconds() / 60, "Minuten")
````

Die Differenz zweier Zeitpunkte ist ein `timedelta`. Auch das Mitternachtsproblem aus LE01 lässt sich damit lesbar lösen: Liegt das Ende vor dem Beginn, addiert man einen Tag.

````python
beginn = datetime.fromisoformat("2026-10-09T22:30")
ende = datetime.fromisoformat("2026-10-09T00:10")
if ende < beginn:
    ende += timedelta(days=1)
print(ende - beginn)
````

💡 `strftime("%A")` liefert Wochentage je nach Systemeinstellung auf Englisch. Für deutsche Kürzel ist eine eigene Liste zuverlässiger: `["Mo", "Di", "Mi", "Do", "Fr", "Sa", "So"][tag.weekday()]`. `weekday()` zählt ab `0` = Montag, `isoweekday()` ab `1`.

### Zeitzonen und UTC

Viele Geräte und Dienste speichern Zeitstempel in **UTC** (koordinierte Weltzeit), erkennbar am `Z` oder `+00:00` am Ende. Deutschland liegt im Winter eine Stunde (MEZ), im Sommer zwei Stunden (MESZ) vor UTC. Mit dem Modul `zoneinfo` rechnen Sie korrekt um, einschließlich Sommerzeit:

````python
from zoneinfo import ZoneInfo

aufwachen_utc = datetime.fromisoformat(schlaf["sleepSessions"][0]["end"])
aufwachen_lokal = aufwachen_utc.astimezone(ZoneInfo("Europe/Berlin"))

print(aufwachen_utc)
print(aufwachen_lokal)
````

⚠️ Am 25.10.2026 endet die Sommerzeit. Wer UTC-Zeiten einfach mit „+2 Stunden“ umrechnet, liegt danach eine Stunde daneben. Das wird in LE06 wichtig, wenn Sie Schlaf- und Lerndaten zeitlich verknüpfen.

## 10 Fehler abfangen
---

Selbst erhobene Daten enthalten fast immer Werte, die sich nicht umwandeln lassen: leere Zellen, Tippfehler, Text statt Zahl. Ohne Vorkehrung bricht Ihr Programm beim ersten solchen Wert ab. Mit `try`/`except` fangen Sie einen **erwarteten** Fehler gezielt ab:

````python
roh = ["4", "", "3", "hoch", "5"]
fokus = []
probleme = []

for wert in roh:
    try:
        fokus.append(int(wert))
    except ValueError:
        probleme.append(wert)

print("gültig:", fokus)
print("nicht umwandelbar:", probleme)
````

Zwei Regeln für den Umgang mit Fehlern in der Datenanalyse:

1. **Fangen Sie nur den erwarteten Fehlertyp ab** (hier `ValueError`), niemals pauschal mit `except:`. Sonst verschleiern Sie echte Programmierfehler.
2. **Verschlucken Sie Probleme nicht stillschweigend.** Sammeln und melden Sie sie wie oben. Ob ein ungültiger Wert fehlt, korrigiert oder entfernt wird, ist eine bewusste Entscheidung, die Sie dokumentieren (LE05).
