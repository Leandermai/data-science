## Teil 1: Zwischenübung Fahrradzählstelle
---

Eine Zählstelle an einem Radweg hat eine Woche lang, von Montag, 2. November, bis Sonntag, 8. November 2026, **jede Stunde** die vorbeifahrenden Räder gezählt. Beantworten Sie die sechs Fragen mit Code. Jede Antwort ist **eine Zahl oder ein Wort**. Wir sammeln die Antworten anschließend im Plenum.

Verwenden Sie **keine Schleifen**.

```python
import numpy as np
import pandas as pd

TAGE = np.array(["Mo", "Di", "Mi", "Do", "Fr", "Sa", "So"])

df = pd.read_csv("data/fahrradzaehlung_2026-11-02.csv")
werte = df["anzahl"].to_numpy()

print(werte.shape)
werte[:12]                      # Montag, 0 bis 11 Uhr
```

### a) Umformen

Die 168 Werte liegen Stunde für Stunde hintereinander: zuerst Montag 0 bis 23 Uhr, dann Dienstag und so weiter. Formen Sie `werte` in ein Array `zaehlung` mit **einer Zeile je Tag** und **einer Spalte je Stunde** um.

**Frage:** Welche Zahl steht in `zaehlung[1, 0]`?

<details><summary>💡 Tipp</summary>

`reshape(zeilen, spalten)`: Eine Woche hat 7 Tage, ein Tag 24 Stunden. Zur Kontrolle: `zaehlung.shape` muss `(7, 24)` ergeben.

</details>

```python
zaehlung = werte.reshape(7, 24)
```

### b) Ein einzelner Wert

**Frage:** Wie viele Räder wurden am **Mittwoch** in der Stunde ab **8 Uhr** gezählt?

<details><summary>💡 Tipp</summary>

`zaehlung[zeile, spalte]`, gezählt wird ab 0: Montag ist Zeile 0.

</details>

```python
zaehlung[2, 8]
```

### c) Ein Ausschnitt

**Frage:** Wie viele Räder wurden an den **fünf Werktagen** in den Stunden ab **7, 8 und 9 Uhr** insgesamt gezählt?

<details><summary>💡 Tipp</summary>

Wählen Sie mit zwei Slices aus: Zeilen 0 bis 4, Spalten 7 bis 9. Bei `start:stop` gehört `stop` nicht mehr dazu. Anschließend `.sum()`.

</details>

```python
berufsverkehr = zaehlung[:5, 7:10].sum()
print(berufsverkehr)
```

### d) Eine Maske

**Frage:** In wie vielen Stunden der Woche wurden **mehr als 100 Räder** gezählt?

<details><summary>💡 Tipp</summary>

Ein Vergleich wie `zaehlung > 100` liefert ein boolesches Array. Beim Summieren zählt `True` als 1.

</details>

```python
mask = zaehlung.mask()
```

### e) Summe je Tag

**Frage:** An welchem **Tag** wurden die wenigsten Räder gezählt?

<details><summary>💡 Tipp</summary>

Für einen Wert je Tag werden die Stunden zusammengefasst: `sum(axis=1)`. `argmin()` liefert die Position des kleinsten Werts, `TAGE[position]` den Namen des Tages.

</details>

```python
TAGE[zaehlung.sum(axis=1),argmin()]
```

### f) Durchschnitt je Stunde

Berechnen Sie `mittel_je_stunde`: für **jede der 24 Stunden** die durchschnittliche Anzahl Räder über die sieben Tage. Das sind 24 Werte, je einer pro Spalte. Der Wert an Position 8 ist zum Beispiel der Mittelwert der sieben Zählwerte in der Stunde ab 8 Uhr (Montag bis Sonntag).

**Frage:** In welcher **Stunde** fahren im Durchschnitt der sieben Tage die meisten Räder?

<details><summary>💡 Tipp</summary>

Für einen Wert je Stunde werden die Tage zusammengefasst: `mean(axis=0)`. `argmax()` liefert die Position des größten Werts, die Position ist hier zugleich die Stunde.

</details>

```python
mittel_je_stunde = zaehlung.mean(axis=0).argmax()
```

---
**Ausblick:** Wie stark weicht jeder Tag in jeder Stunde vom Durchschnitt dieser Stunde ab? Dafür müsste `zaehlung` mit der Form `(7, 24)` mit `mittel_je_stunde` der Form `(24,)` verrechnet werden, zwei Arrays unterschiedlicher Form. Wie das geht, zeigt der nächste Abschnitt: **Broadcasting**.
