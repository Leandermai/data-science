#### Fragen, die Ideen auslösen

* **Was mache ich regelmäßig?** Alles, was täglich oder mehrmals pro Woche passiert, liefert viele Beobachtungen.
* **Worüber stelle ich Behauptungen auf?** Sätze wie „Montags ist der Bus immer zu spät“ oder „Abends spiele ich schlechter“ sind prüfbare Hypothesen.
* **Was nervt mich, oder was würde ich gern verbessern?** Wer ein echtes Problem lösen will, bleibt motiviert.
* **Welche Apps oder Geräte sammeln schon Daten über mich?** Vieles lässt sich exportieren, wird aber oft nie angeschaut.
* **Worüber könnte ich stundenlang erzählen?** Dort haben Sie das Domänenwissen, das gute Analysen braucht.
* **Welche Entscheidung steht an?** Daten helfen z. B. bei der Wahl zwischen Rad und Bus oder bei der besten Uhrzeit fürs Fitnessstudio.
* **Kombinieren:** Zwei Bereiche zu verknüpfen, ergibt oft eine gute mehrdimensionale Frage und gleich die zweite Datenquelle, z. B. Sport × Wetter, Lernen × Schlaf oder Musik × Stimmung.

#### Warnsignale

Diese Anzeichen sprechen dafür, dass eine Idee (noch) nicht trägt:

* Die Daten gibt es **nur öffentlich** (Kaggle, Statistikämter). Dann fehlt der eigene Anteil.
* Es entsteht **nur ein Wert pro Tag** über wenige Wochen. Dann sollten Sie auf Ereignisebene erheben (jede Fahrt, jedes Spiel, jede Mahlzeit).
* Die Erhebung ist so **aufwendig**, dass Sie sie nach zwei Wochen aufgeben.
* Die Daten sind **sehr privat** oder betreffen andere Personen ohne deren Einwilligung.
* Die **Antwort steht schon fest**, oder die Frage ist **zu groß** („Wie verändert sich das Klima?“).

Ein Warnsignal ist kein K.-o.-Kriterium. Oft lässt sich eine Idee so umbauen, dass sie trägt.
