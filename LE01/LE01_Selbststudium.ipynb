## 1 Was ist Data Science?
---

**Data Science** ist ein interdisziplinäres Feld, das mit wissenschaftlichen Methoden, Prozessen und Algorithmen **Erkenntnisse aus Daten** gewinnt. Häufig wird Data Science als Schnittmenge dreier Kompetenzbereiche beschrieben (nach Drew Conway, 2010):

* **Programmierung und Informatik:** Daten beschaffen, verarbeiten und automatisieren,
* **Mathematik und Statistik:** Muster erkennen, Unsicherheit quantifizieren, Ergebnisse absichern,
* **Domänenwissen:** die richtigen Fragen stellen und Ergebnisse sinnvoll interpretieren.

Gerade der dritte Punkt wird oft unterschätzt. Ohne Domänenwissen entstehen korrekt berechnete, aber bedeutungslose Zahlen. In Ihrem Semesterprojekt sind Sie selbst die Domänenexpertin oder der Domänenexperte, denn Sie analysieren Daten aus Ihrem eigenen Leben.

### Vier Arten von Analysen

| Analyseart | Typische Frage | Beispiel „Lerntagebuch“ |
|---|---|---|
| **Deskriptiv** | Was ist passiert? | Wie viele Stunden habe ich pro Woche gelernt? |
| **Diagnostisch** | Warum ist es passiert? | Lerne ich an Tagen mit wenig Schlaf kürzer? |
| **Prädiktiv** | Was wird passieren? | Wie viele Lernstunden schaffe ich bis zur Klausur? |
| **Präskriptiv** | Was sollten wir tun? | Zu welcher Tageszeit sollte ich Mathe lernen? |

In diesem Kurs konzentrieren wir uns auf **deskriptive und diagnostische (explorative) Analysen** und auf die **Kommunikation** der Ergebnisse. Prädiktive Modelle mit Machine Learning sind Thema anderer Module, z. B. *Künstliche Intelligenz*. Sie profitieren aber stark von den Fertigkeiten, die Sie hier erwerben. Ein Großteil der Zeit in echten Data-Science-Projekten fließt nämlich in das Beschaffen, Verstehen und Aufbereiten von Daten.

### Warum Python?

Python hat sich als Standardsprache für Data Science etabliert. Die Sprache selbst ist leicht zu lesen, und um sie herum ist ein umfangreiches Ökosystem an Bibliotheken entstanden, der **PyData-Stack**. Diese Werkzeuge lernen Sie im Laufe des Semesters kennen:

## 2 CRISP-DM: ein Vorgehensmodell für Datenprojekte
---

Datenprojekte scheitern selten an der Technik, sondern häufiger daran, dass unklar ist, **welche Frage** eigentlich beantwortet werden soll, oder dass die Daten diese Frage gar nicht beantworten können. Ein Vorgehensmodell hilft, strukturiert zu arbeiten.

Das bekannteste Modell ist **CRISP-DM** (*Cross-Industry Standard Process for Data Mining*). Es wurde Ende der 1990er Jahre von einem Industriekonsortium entwickelt und ist bis heute der Standard in der Praxis. CRISP-DM passt nicht nur für Data Mining im engeren Sinne, sondern für praktisch jedes datengetriebene Projekt. **Ihr Semesterprojekt strukturieren Sie nach CRISP-DM.**

<img src="img/crisp-dm.png" width="480" alt="CRISP-DM-Prozessmodell mit sechs Phasen um die Daten herum">

<small>Abbildung: Kenneth Jensen, [CRISP-DM Process Diagram](https://commons.wikimedia.org/wiki/File:CRISP-DM_Process_Diagram.png), [CC BY-SA 3.0](https://creativecommons.org/licenses/by-sa/3.0/), via Wikimedia Commons</small>

Zwei Eigenschaften sind zentral:

* **Iterativ:** Die Pfeile zwischen den Phasen zeigen, dass man häufig zurückspringt. Beim Aufbereiten der Daten stellt man etwa fest, dass eine Variable fehlt, und muss zurück zur Datenerhebung.
* **Zyklisch:** Der äußere Kreis zeigt, dass ein Projekt nicht mit dem Deployment endet. Die Ergebnisse werfen in der Regel neue Fragen auf.

### Die sechs Phasen

Da Sie in diesem Kurs keine Machine-Learning-Modelle trainieren, interpretieren wir die Phase *Modeling* als **Analyse**: Kennzahlen berechnen, gruppieren, pivotisieren, Zusammenhänge untersuchen.

| Phase | Typische Frage | Typische Tätigkeiten | Ergebnis im Semesterprojekt | LE |
|---|---|---|---|---|
| **Business Understanding** | Was will ich wissen und warum? | Ziel und Kontext klären, Leitfragen formulieren, Erfolgskriterien festlegen, Datenquellen planen | Projektsteckbrief | 01 |
| **Data Understanding** | Was steckt in meinen Daten? | Daten erheben, beschreiben und erkunden, Datenqualität prüfen | Data Dictionary, erste Übersicht | 02–03 |
| **Data Preparation** | Wie werden die Daten analysierbar? | Auswählen, bereinigen, verknüpfen, umformen, neue Merkmale ableiten | Bereinigter, *tidy* Datensatz | 04–07 |
| **Modeling** (bei uns: **Analyse**) | Was sagen die Daten zu meiner Frage? | Deskriptive Statistik, Gruppierungen, Pivottabellen, Zusammenhänge, einfache Tests | Analyse-Notebook | 08–09 |
| **Evaluation** | Beantworten die Ergebnisse meine Frage? | Plausibilität prüfen, Grenzen und Verzerrungen benennen, nächste Iteration planen | Kritische Bewertung | 08–10 |
| **Deployment** | Wie nutzen andere meine Ergebnisse? | Visualisieren, Data Story entwickeln, präsentieren, Bericht erstellen | Projektdokument und Präsentation | 10–12 |

Die Veranstaltung selbst ist entlang dieser Phasen aufgebaut:

* **Phase A – Business/Data Understanding** (LE01–LE03): Leitfragen, Datenquellen, Datenstrukturen, erste explorative Analyse
* **Phase B – Data Preparation** (LE03–LE07): pandas, NumPy, Vektorisierung, Clean Data, Merging, Tidy Data
* **Phase C – Exploratory Data Analysis** (LE08–LE10): deskriptive Statistik, Gruppierungen, OLAP, Pivotisierung, Visualisierung
* **Phase D – Analysis** (LE08–LE09): statistische Kennzahlen, Zusammenhänge, einfache statistische Tests
* **Phase E – Communication** (LE10–LE12): effektive und interaktive Visualisierung, Storytelling

### Gute Leitfragen formulieren

Die Fragen, die Sie mit Ihrem Projekt beantworten wollen, nennen wir im Kurs **Leitfragen**. Sie sind das Fundament Ihres Projekts: Sie stehen in Abschnitt 2 Ihres Projektsteckbriefs, werden laufend geschärft und begleiten Sie durch das ganze Semester. Sie können mehrere Leitfragen bearbeiten, die zusammen Ihre Data Story ergeben, etwa eine zu Ihrer Kennzahl und eine zu möglichen Einflussfaktoren. Meist sind ein bis drei Leitfragen gut zu schaffen. Beantworten Sie lieber wenige Fragen gründlich als viele oberflächlich.

Eine gute datengetriebene Leitfrage ist …

1. **konkret:** Man erkennt, *was* gemessen wird und *wonach* unterschieden wird,
2. **mit Daten beantwortbar,** die Sie realistisch selbst erheben können,
3. **mehrdimensional:** Sie betrachtet mindestens zwei Dimensionen,
4. **relevant:** Sie interessiert Sie selbst, und Sie können Ihrem Publikum erklären, warum die Antwort wichtig ist,
5. **offen:** Die Antwort ist nicht schon vorher klar und nicht bloß ein „Ja“ oder „Nein“.

Für Kriterium 3 hilft eine Unterscheidung, die uns in LE09 (OLAP und Pivottabellen) wieder begegnet:

* Eine **Kennzahl** (*Measure*, *Fakt*) ist die Größe, die gemessen und verrechnet wird, z. B. *Lernzeit in Minuten*.
* Eine **Dimension** ist ein Merkmal, *nach dem* aufgeteilt oder gruppiert wird, z. B. *Wochentag* oder *Fach*.

In der Frage *„Wie unterscheidet sich meine **Lernzeit** nach **Wochentag** und **Fach**?“* ist die Lernzeit die Kennzahl, Wochentag und Fach sind die Dimensionen.

| Schwache Frage | Problem | Bessere Frage |
|---|---|---|
| Wie ist mein Schlaf? | Unkonkret, keine Dimension | Wie hängt meine Schlafdauer mit der Einschlafzeit und dem Wochentag zusammen, und unterscheiden sich Vorlesungs- und Wochenendtage? |
| Was höre ich auf Spotify? | Rein beschreibend, nur eine Dimension | Wie unterscheiden sich meine gehörten Genres nach Tageszeit und Aktivität (Lernen, Sport, Pendeln)? |
| Ist die Mensa gut? | „Gut“ ist nicht messbar | Wie hängt die Wartezeit in der Mensa von Wochentag und Uhrzeit ab? |
| Wie viel Geld gebe ich für Kaffee aus? | Eine einzige Zahl ist schnell ermittelt | Wofür gebe ich wann Geld aus, nach Kategorie, Wochentag und Zahlungsart? |

## 3 Setup: Python mit uv
---

In der Praxis arbeitet man nie „einfach mit Python“, sondern immer in einer **Projektumgebung**: einem Ordner mit einer festgelegten Python-Version und genau den Paketen, die das Projekt braucht. Dafür verwenden wir das Werkzeug [**uv**](https://docs.astral.sh/uv/). Es installiert Python-Versionen, verwaltet Pakete und legt pro Projekt eine isolierte, virtuelle Umgebung an (Ordner `.venv`).

Warum ist das gerade in Data Science wichtig? Analysen müssen **reproduzierbar** sein. Wenn Ihr Notebook heute mit pandas 3.0 läuft, soll es das auch bei Ihrer Prüferin oder Ihrem Prüfer tun. uv hält dafür in zwei Dateien fest, was ein Projekt braucht:

* `pyproject.toml`: die Projektbeschreibung, also welche Pakete ungefähr benötigt werden, z. B. `pandas>=3.0`,
* `uv.lock`: die **exakten** Versionen aller Pakete und ihrer Abhängigkeiten.

### Schritt für Schritt

1. **uv und Python installieren:** Folgen Sie der Anleitung *„Python mit uv installieren“* auf Moodle (Linux). Für Windows und macOS finden Sie die Installationsbefehle in der [uv-Dokumentation](https://docs.astral.sh/uv/getting-started/installation/). Danach installieren Sie Python mit `uv python install 3.14`.
2. **Kursmaterial einrichten:** Das Kursmaterial ist selbst ein uv-Projekt. Entpacken bzw. klonen Sie es, wechseln Sie im Terminal in den Ordner und führen Sie einmalig aus:
   ```bash
   uv sync
   ```
   uv installiert dann genau die Python-Version und die Pakete, die für den Kurs festgelegt sind.
3. **Notebooks öffnen:**
   * **JupyterLab im Browser:** Im Kursordner `uv run jupyter lab` ausführen.
   * **VS Code:** Den Kursordner öffnen, die Erweiterungen *Python* und *Jupyter* installieren und oben rechts im Notebook den Kernel aus `.venv` auswählen.

Ihr **eigenes Semesterprojekt** legen Sie in der Präsenz als separates uv-Projekt an (`uv init`). Kursmaterial und Projekt bleiben so sauber getrennt.

### Die wichtigsten uv-Befehle

| Befehl | Wirkung |
|---|---|
| `uv init --no-package name` | Neues Projekt im Ordner `name` anlegen |
| `uv sync` | Umgebung eines vorhandenen Projekts herstellen |
| `uv add paket` | Paket zum Projekt hinzufügen (statt `pip install`) |
| `uv remove paket` | Paket entfernen |
| `uv run datei.py` | Skript in der Projektumgebung ausführen |
| `uv run python` | Interaktive Python-Konsole des Projekts starten |
| `uv run jupyter lab` | JupyterLab in der Projektumgebung starten |

> ⚠️ Installieren Sie Pakete **nicht** mit `pip install` oder `!pip install` im Notebook, sondern im Terminal mit `uv add`. Nur so landet das Paket in `pyproject.toml` und `uv.lock`, und Ihr Projekt bleibt reproduzierbar.

## 4 Arbeiten mit Jupyter Notebooks
---

Ein **Jupyter Notebook** (Dateiendung `.ipynb`) kombiniert ausführbaren Code, formatierten Text, Formeln und Grafiken in einem Dokument. Genau das macht Notebooks in Data Science so beliebt: Sie können explorativ arbeiten und Ihre Analyse gleichzeitig dokumentieren. Auch Ihr Projektdokument wird aus Notebooks bestehen.

Ein Notebook besteht aus **Zellen**:

* **Code-Zellen** enthalten Python-Code. Die Ausgabe erscheint direkt unter der Zelle.
* **Markdown-Zellen** enthalten formatierten Text, so wie diese Zelle.

Hinter dem Notebook läuft ein **Kernel**, also ein Python-Prozess. Er führt den Code aus und merkt sich alle Variablen. Die Nummer links neben einer Code-Zelle, z. B. `[5]`, zeigt, als wievielte Zelle sie ausgeführt wurde.

### Wichtige Tastenkürzel (JupyterLab)

Jupyter kennt zwei Modi. Im **Bearbeitungsmodus** (`Enter`) schreiben Sie in eine Zelle, im **Befehlsmodus** (`Esc`) arbeiten Sie mit ganzen Zellen.

| Kürzel | Modus | Wirkung |
|---|---|---|
| `Shift + Enter` | beide | Zelle ausführen und zur nächsten springen |
| `Ctrl + Enter` | beide | Zelle ausführen und stehen bleiben |
| `Esc` / `Enter` | – | In den Befehls- bzw. Bearbeitungsmodus wechseln |
| `A` / `B` | Befehl | Neue Zelle oberhalb (*above*) bzw. unterhalb (*below*) einfügen |
| `M` / `Y` | Befehl | Zelle in Markdown bzw. Code umwandeln |
| `D, D` | Befehl | Zelle löschen |
| `Z` | Befehl | Löschen rückgängig machen |
| `0, 0` | Befehl | Kernel neu starten |
| `Tab` | Bearbeiten | Autovervollständigung |
| `Shift + Tab` | Bearbeiten | Dokumentation zur Funktion anzeigen |

### Achtung: Ausführungsreihenfolge

Der Kernel kennt nur, was Sie **tatsächlich ausgeführt** haben, und zwar in der Reihenfolge, in der Sie es ausgeführt haben. Die Position im Dokument spielt keine Rolle. Das ist praktisch, aber auch die häufigste Fehlerquelle in Notebooks.

✏️ **Ausprobieren:** Führen Sie die erste der beiden folgenden Zellen einmal aus, die zweite dann **mehrmals**. Beobachten Sie die Ausgabe und die Nummern links neben den Zellen.

````python
zaehler = 0
````

````python
zaehler = zaehler + 1
zaehler
````

Die zweite Zelle liefert bei jeder Ausführung ein anderes Ergebnis, obwohl der Code gleich bleibt. Noch tückischer: Löschen Sie eine Zelle, existieren ihre Variablen im Kernel weiter. Ihr Notebook funktioniert dann bei Ihnen, aber nicht mehr bei anderen.

> 💡 **Best Practice:** Starten Sie vor dem Speichern oder Abgeben den Kernel neu und führen Sie alle Zellen von oben nach unten aus (JupyterLab: *Kernel → Restart Kernel and Run All Cells…*, VS Code: *Restart* und danach *Run All*). Ein Notebook gilt erst dann als fertig, wenn es von oben nach unten fehlerfrei durchläuft.

### 5.4 Strings

Strings (`str`) schreiben Sie mit einfachen oder doppelten Anführungszeichen. Mehrzeilige Strings stehen in dreifachen Anführungszeichen. Strings sind **unveränderlich** (*immutable*): Jede Methode liefert einen neuen String zurück.

**Indexing und Slicing:** Der Index beginnt bei `0`, negative Indizes zählen von hinten. Mit `[start:stop:schritt]` schneiden Sie Teilstrings aus, wobei `stop` nicht mehr dazugehört.

````python
datum = "2026-10-05"
print(len(datum))
print(datum[0])      # erstes Zeichen
print(datum[-1])     # letztes Zeichen
print(datum[0:4])    # Zeichen 0 bis 3 -> Jahr
print(datum[5:7])    # Monat
print(datum[::-1])   # rückwärts
````

**Methoden** rufen Sie mit Punktnotation auf. Für die Datenbereinigung sind diese besonders wichtig:

| Methode | Wirkung | Beispiel | Ergebnis |
|---|---|---|---|
| `strip()` | Leerzeichen am Rand entfernen | `"  Mathe ".strip()` | `'Mathe'` |
| `lower()`, `upper()`, `title()` | Groß-/Kleinschreibung | `"mathe".title()` | `'Mathe'` |
| `replace(alt, neu)` | Ersetzen | `"3,5".replace(",", ".")` | `'3.5'` |
| `split(trenner)` | Zerlegen in eine Liste | `"a;b;c".split(";")` | `['a', 'b', 'c']` |
| `startswith(...)` | Prüft den Anfang | `"#Kommentar".startswith("#")` | `True` |
| `count(...)` | Vorkommen zählen | `"Mississippi".count("ss")` | `2` |

Mit `in` prüfen Sie, ob ein Teilstring enthalten ist. Methoden lassen sich verketten:

### 5.5 f-Strings: formatierte Ausgabe

Mit **f-Strings** betten Sie Variablen und Ausdrücke in einen Text ein. Dazu stellen Sie ein `f` vor den String und schreiben die Ausdrücke in `{}`. Nach einem Doppelpunkt können Sie das Format genauer festlegen ([Format Specification Mini-Language](https://docs.python.org/3/library/string.html#formatspec)).

````python
fach = "Statistik"
dauer_min = 95
anteil = 0.2567
umsatz = 1234567.891

print(f"{fach}: {dauer_min} min")
print(f"{fach}: {dauer_min / 60:.1f} h")      # 1 Nachkommastelle
print(f"Anteil: {anteil:.1%}")                # Prozent
print(f"Umsatz: {umsatz:,.2f} €")             # Tausendertrenner, 2 Stellen
print(f"|{fach:<12}|{dauer_min:>6}|")         # links- und rechtsbündig in fester Breite
print(f"{dauer_min=}")                        # praktisch zum Debuggen
````

| Format | Bedeutung | Beispiel mit `x = 3.14159` | Ergebnis |
|---|---|---|---|
| `:.2f` | Festkomma mit 2 Nachkommastellen | `f"{x:.2f}"` | `3.14` |
| `:.1%` | Prozent mit 1 Nachkommastelle | `f"{0.256:.1%}"` | `25.6%` |
| `:,` | Tausendertrenner | `f"{1234567:,}"` | `1,234,567` |
| `:>8`, `:<8`, `:^8` | rechts-, linksbündig, zentriert (Breite 8) | `f"{'ab':>8}"` | `'      ab'` |
| `:05d` | mit Nullen auffüllen | `f"{7:05d}"` | `00007` |

💡 f-Strings verwenden die englische Schreibweise mit Dezimalpunkt. Für deutsche Zahlenformate hilft z. B. `f"{x:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".")`. Für Grafiken lernen Sie später elegantere Wege kennen.

### 5.6 Typumwandlung

Mit `int()`, `float()`, `str()` und `bool()` wandeln Sie Werte um. Das ist in Data Science ständig nötig, denn **alles, was Sie aus einer Textdatei lesen, ist zunächst ein String**.

````python
print(int("42") + 1)
print(float("3.5") * 2)
print(str(42) + " Minuten")
print(int(3.9))          # schneidet ab, rundet nicht!
print(round(3.9))
print(round(2.5), round(3.5))   # rundet auf die nächste gerade Zahl
````

`round()` rundet bei `.5` auf die nächste **gerade** Zahl (*Banker's Rounding*), damit sich Rundungsfehler bei vielen Werten nicht systematisch aufsummieren.

Und was passiert mit einer deutschen Dezimalzahl?

````python
float("3,5")
````

Die Umwandlung schlägt mit einem `ValueError` fehl. Solche Formatprobleme sind typisch für selbst erhobene Daten. Eine einfache Lösung:

````python
float("3,5".replace(",", "."))
````

### 5.7 Verzweigungen

Blöcke werden in Python durch **Einrückung** gebildet, die Zeile davor endet mit einem Doppelpunkt. Statt `else if` schreibt man `elif`.

### 5.8 Schleifen

Die `for`-Schleife in Python läuft immer **über die Elemente einer Folge**, z. B. über die Zeichen eines Strings, die Elemente einer Liste oder die Zahlen aus `range()`. Eine Zählschleife im Stil von `for (int i = 0; i < 5; i++)` schreiben Sie mit `range`:

````python
for i in range(5):          # 0, 1, 2, 3, 4
    print(i, end=" ")
print()
for i in range(10, 0, -3):  # start, stop (exklusiv), schritt
    print(i, end=" ")
````

Typisch für die Datenverarbeitung ist das **Akkumulieren** in einer Schleife. Hier verarbeiten wir ein kleines Lernprotokoll, das als mehrzeiliger String vorliegt. `splitlines()` zerlegt ihn in einzelne Zeilen:

````python
protokoll = """Mathe;45
Statistik;90
Mathe;30
Programmieren;120"""

gesamt = 0
for zeile in protokoll.splitlines():
    fach, dauer = zeile.split(";")
    gesamt += int(dauer)
    print(f"{fach:<15}{dauer:>5} min")

print("-" * 25)
print(f"{'Gesamt':<15}{gesamt:>5} min")
````

`while`-Schleifen laufen, solange eine Bedingung gilt. `break` beendet eine Schleife vorzeitig, `continue` springt zum nächsten Durchlauf.

````python
ziel_min = 300
woche = 0
kumuliert = 0
while kumuliert < ziel_min:
    woche += 1
    kumuliert += 70
print(f"Ziel nach {woche} Wochen erreicht ({kumuliert} min).")
````
